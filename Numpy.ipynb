```python
def DataEntry():
    import numpy as np;
    arr = [];
    n = int(input("Entry array size: "));
    for i in range(n):
        val = int(input(f"Enter element {i + 1}: "))
        arr.append(val);
    search = int(input("Enter the number you want to search: "))

    if search in arr:
        index = arr.index(search)
        print(f"{search} found in the array at position {index + 1}");
    else:
        print(f"{search} not found in the array.");
DataEntry();
```
